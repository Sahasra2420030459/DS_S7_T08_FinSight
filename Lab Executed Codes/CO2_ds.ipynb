```python
#Data inspection
import pandas as pd
from sklearn.datasets import load_iris
# Load iris dataset
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['species'] = iris.target
# Map species numbers to names
df['species'] = df['species'].map({0:'setosa', 1:'versicolor', 2:'virginica'})
# Inspect
print(df.head())
print(df.info())
print(df.isnull().sum())
```

```python
# Categorical summary
print(df.describe())
print(df['species'].value_counts())
```

```python
#Data Understanding

import seaborn as sns
import matplotlib.pyplot as plt
sns.pairplot(df, hue="species")
plt.show()
#Pair Plot():
sns.boxplot(x="species", y="petal length (cm)", data=df)
plt.show()
#BoxPlot():
sns.heatmap(df.drop(columns="species").corr(), annot=True, cmap="coolwarm")
plt.show()
```

```python
# IQR Calculation

import pandas as pd
import numpy as np

# Create the dataset
df = pd.DataFrame({
    'data science': [10, 12, 15, 18, 20, 22, 25, 27, 30, 35, 100]
})

print("Dataset:")
print(df)

# Q1
Q1 = np.percentile(df['data science'], 25, method='midpoint')

# Q3
Q3 = np.percentile(df['data science'], 75, method='midpoint')

# IQR
IQR = Q3 - Q1

print("\nThe output of IQR:")
print(IQR)

# Upper bound
upper = Q3 + 1.5 * IQR

upper_array = np.array(df['data science'] >= upper)

print("\nUpper Bound:", upper)
print("Number of upper outliers:", upper_array.sum())

# Lower bound
lower = Q1 - 1.5 * IQR

lower_array = np.array(df['data science'] <= lower)

print("Lower Bound:", lower)
print("Number of lower outliers:", lower_array.sum())
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
# Load Dataset
iris = sns.load_dataset("iris")
# Pair Plot
sns.pairplot(
    data=iris,                  # Dataset
    hue="species",              # Color by species
    hue_order=["setosa", "versicolor", "virginica"],
    palette="Set1",             # Color palette
    vars=[
        "sepal_length",
        "sepal_width",
        "petal_length",
        "petal_width"
    ],
    kind="scatter",             # Scatter plots
    diag_kind="hist",           # Histogram on diagonal
    markers=["o", "s", "D"],    # Different markers
    height=3,                   # Height of each plot
    aspect=1.2,                 # Width/Height ratio
    corner=False,               # Show full matrix
    dropna=True                 # Remove missing values
)
plt.show()
```

```python
# Callbacks

import dash
from dash import dcc, html, Input, Output

# Create Dash application
app = dash.Dash(__name__)

# Define layout
app.layout = html.Div([

    html.H1("Square Calculator"),

    dcc.Input(
        id='input',
        type='text',
        placeholder='Enter a number'
    ),

    # Output area
    html.Div(
        id='output'
    )
])

# Callback
@app.callback(
    Output('output', 'children'),
    Input('input', 'value')
)
def calculate_square(value):
    if value is None or value == '':
        return "Enter a number"

    try:
        number = float(value)
        return f"Square = {number ** 2}"
    except ValueError:
        return "Please enter a valid number"

# Run application
app.run(debug=True, port=8051)
```
